## 3. Success metric

*One metric you can defend. What number means 'good'?*

A defensible success metric for clustering is the average silhouette score across all content items — a score above 0.5 means clusters are well-separated and internally cohesive; below 0.2 means the clustering isn't meaningfully distinguishing archetypes.

```python
from sklearn.metrics import silhouette_score
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed if-statement rule can't capture this because "successful content" isn't defined by a single threshold — it depends on interacting patterns across many features at once (search volume, competition, word count, content type, intent). There's no obvious cutoff like "word_count > 3000" that separates good archetypes from bad ones; the relationships between features matter more than any single value. ML lets these multi-dimensional patterns emerge from the data instead of me guessing arbitrary rule thresholds.

```python
# Example: no single fixed threshold captures 'successful' content —
# it depends on combinations across features, which clustering can detect
# but a simple rule (e.g. word_count > 3000) cannot.
print(df[['word_count', 'search_volume', 'competition', 'content_type']].describe())
```

A wrong call costs wasted editor hours — reworking content that didn't actually need it, or missing content that did. A plain rule isn't enough because "successful content" depends on multi-dimensional, interacting patterns, not a single fixed threshold.
